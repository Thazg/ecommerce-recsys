# Milestone 2 — Data Understanding and EDA

This notebook analyzes the Retailrocket raw dataset to understand:
- Behavioral funnel distribution (view, addtocart, transaction)
- User activity sparsity and cold-start proportions
- Item popularity and long-tail decay
- Repeated interactions and duplicate behaviors


```python
from pathlib import Path

import pandas as pd

raw_events_path = Path("../data/raw/events.csv")
df = pd.read_csv(raw_events_path)
print(f"Total events loaded: {len(df):,}")
df.head()
```

```python
# Event Type Breakdown
df["event"].value_counts(normalize=True).mul(100).round(2)
```

```python
# User Activity Distribution
user_counts = df.groupby("visitorid").size()
print(f"Proportion of users with exactly 1 interaction: {(user_counts == 1).mean():.2%}")
user_counts.quantile([0.5, 0.75, 0.9, 0.95, 0.99])
```
